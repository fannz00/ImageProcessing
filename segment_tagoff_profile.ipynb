```python
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
from skimage import filters, measure, morphology, exposure

# 1. Load images from a folder, sorted by timestamp in filename
folder = '/home/veit/Documents/20250717_PISCO-SCHARF/20250716-1423_PISCO_test_01_PISCO2_png'

def extract_timestamp(fname):
    # Assumes timestamp is the last field before the extension, separated by '_'
    base = os.path.splitext(fname)[0]
    parts = base.split('_')
    return parts[-1]  # e.g., '20250716-14320885'

# List and sort files by extracted timestamp
file_list = [f for f in os.listdir(folder) if f.endswith('.tif') or f.endswith('.png')]
file_list = sorted(file_list, key=extract_timestamp)

images = []
for fname in file_list:
    img = cv2.imread(os.path.join(folder, fname), cv2.IMREAD_GRAYSCALE)
    images.append(img)
images = np.array(images)
```
